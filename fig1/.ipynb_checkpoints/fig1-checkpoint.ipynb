```python
%autosave 90
%config Completer.use_jedi = False
from IPython.core.interactiveshell import InteractiveShell
from IPython.display import HTML

InteractiveShell.ast_node_interactivity = "all"
import glob
import pickle

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from adjustText import adjust_text
from scipy.stats import kstest, mannwhitneyu, ranksums, ttest_ind
from statsmodels.stats.multitest import fdrcorrection, multipletests

%config InlineBackend.figure_format = 'retina'
from tqdm import tqdm
```

```python
C9_patient_carrier_DMR_annotation = pd.read_csv(
    "./20241002_denovo_DMR_C9ALS_vs_carrier_promoter_recalculate_MWU_pvalue.tsv",
    sep="\t",
)
C9_patient_sALS_DMR_annotation = pd.read_csv(
    "./20241002_denovo_DMR_C9ALS_vs_sALS_promoter_recalculate_MWU_pvalue.tsv",
    sep="\t",
)
C9_patient_ctrl_DMR_annotation = pd.read_csv(
    "./20241002_denovo_DMR_C9ALS_vs_control_promoter_recalculate_MWU_pvalue.tsv",
    sep="\t",
)
sALS_ctrl_DMR_annotation = pd.read_csv(
    "./20241002_denovo_DMR_sALS_vs_ctrl_promoter_recalculate_MWU_pvalue.tsv",
    sep="\t",
)
sALS_carrier_DMR_annotation = pd.read_csv(
    "./20241002_denovo_DMR_sALS_vs_carrier_promoter_recalculate_MWU_pvalue.tsv",
    sep="\t",
)
```

```python
import matplotlib as mpl
mpl.rcParams['font.family'] = 'Arial'
```

```python
i = 3
for i in range(4):
    group_A, group_B = [
        ("C9-Patient", "C9-Carrier"),
        ("C9-Patient", "sALS"),
        ("C9-Patient", "Control"),
        ("sALS", "Control"),
    ][i]
    plot_temp = pd.read_csv(
        f"./20241002_denovo_DMR_{group_A}_vs_{group_B}_promoter_recalculate_MWU_pvalue.tsv",
        sep="\t",
    )
    plot_temp["abs(diff)"] = plot_temp["mean difference"].abs()
    mpl.rcParams["font.family"] = "Arial"
    sns.set(context="paper", style="ticks")
    
    f, ax = plt.subplots(figsize=(3.35, 2.5))
    ax.set_xlim(-0.3, 0.3)
    # ax.set_ylim(-2, 100)
    sns.scatterplot(
        plot_temp,
        x="mean difference",
        y="-log10(q)",
        ax=ax,
        hue="hue",
        size=5,
        alpha=0.5,
        linewidth=0.1,
        palette={1: "#d62828", 0: "#9b9b98", -1: "#003049"},
    )
    
    ax.axhline(-np.log10(0.00001), zorder=0, c="#9b9b98", lw=0.6, ls="--", alpha=0.5)
    ax.axvline(0.1, zorder=0, c="#d62828", lw=0.6, ls="--", alpha=0.5)
    ax.axvline(-0.1, zorder=0, c="#003049", lw=0.6, ls="--", alpha=0.5)
    
    # ax.set(box_aspect=)
    ax.legend_ = None
    ax.set(
        title="%s - %s" % (group_A, group_B),
        ylabel=r"$-\log_{10}(\mathrm{adjusted\ p\text{-}value})$",
        xlabel="Mean difference",
    )
    ax.title.set_fontsize(8)
    ax.xaxis.label.set_fontsize(8)
    ax.yaxis.label.set_fontsize(8)
    # optional: tick labels too
    ax.tick_params(axis="both", labelsize=7)
    
    # for i, point in C9_plot_temp.head(5).iterrows():
    #    ax.text(point['mean difference']-.1, point['-log10(q)']+2, str(point['annot.symbol']),size=10)
    
    texts = []
    for i, point in (
        plot_temp[
            (plot_temp["-log10(q)"] > 8)
            & (~plot_temp["annot.symbol"].str.startswith(("MIR", "LOC", "LINC")))
            & (~plot_temp["annot.symbol"].str.contains("AS"))
        ]
        .sort_values(by="abs(diff)", ascending=False)[
            ["mean difference", "-log10(q)", "annot.symbol"]
        ]
        .drop_duplicates()
        .head(15)
        .iterrows()
    ):
        if point["mean difference"] > 0:
            ha = "left"
        else:
            ha = "right"
        texts.append(
            ax.text(
                point["mean difference"],
                point["-log10(q)"],
                str(point["annot.symbol"]),
                size=7,
                ha=ha,
            )
        )
    
    
    adjust_text(
        texts,
        arrowprops=dict(arrowstyle="-", color="grey", lw=0.5),
        expand_text=(1.2, 1.4),
        expand_points=(1.1, 1.3),
        force_text=(0.0, 1.5),  # mostly vertical push
        force_points=(0.3, 0.6),
        lim=20000,
        prevent_crossings=True,
        only_move={"text": "y"},  # key: prevent drifting across x=0
    )
    
    f.savefig(f"./{group_A}_{group_B}_recalculate_MWUp_volvano.svg", bbox_inches="tight")
```

```python
mpl.rcParams['font.family'] = 'Arial'
sns.set(context="paper", style="ticks")
for i in range(5):
    df_ = [
        C9_patient_carrier_DMR_annotation,
        C9_patient_sALS_DMR_annotation,
        C9_patient_ctrl_DMR_annotation,
        sALS_ctrl_DMR_annotation,
        sALS_carrier_DMR_annotation,
    ][i]
    plot_temp = df_[
        df_.chr.isin(["chr" + str(_) for _ in list(range(1, 23))])
    ].sort_values(by="chr", ignore_index=True)

    group_A, group_B = [
        ("C9-Patient", "C9-Carrier"),
        ("C9-Patient", "sALS"),
        ("C9-Patient", "Control"),
        ("sALS", "Control"),
        ("sALS", "C9-Carrier"),
    ][i]
    plot_temp = plot_temp[
        (plot_temp["#CpGs"] >= 10)
        & (~plot_temp["p(MWU)_new"].isna())
        & (plot_temp["mean difference"].abs() <= 1)
    ]

    plot_temp["-log10(q)"] = -np.log10(plot_temp["q-value"].values)
    plot_temp["-log10(p)"] = -np.log10(plot_temp["p(MWU)_new"].values)
    plot_temp["q_new"] = fdrcorrection(plot_temp["p(MWU)_new"], method="i")[1]
    plot_temp["-log10(q)"] = -np.log10(plot_temp["q_new"].values)

    plot_temp["hue"] = 0

    plot_temp["hue"][
        (plot_temp["mean difference"] >= 0) & (plot_temp["q_new"] < 0.00001)
    ] = 1
    plot_temp["hue"][
        (plot_temp["mean difference"] <= 0) & (plot_temp["q_new"] < 0.00001)
    ] = -1
    plot_temp[
        [
            "chr",
            "start",
            "stop",
            "mean difference",
            "#CpGs",
            "mean g1",
            "mean g2",
            "annot.symbol",
            "annot.type",
            "p(MWU)_new",
            "-log10(q)",
            "hue",
        ]
    ].reset_index(drop=True).to_csv(
        "./20241002_denovo_DMR_%s_vs_%s_promoter_recalculate_MWU_pvalue.tsv"
        % (group_A, group_B),
        sep="\t",
        index=False,
    )
    # 003049 #9b9b98
    f, ax = plt.subplots(figsize=(3.35, 2.5))
    ax.set_xlim(-0.3, 0.3)
    # ax.set_ylim(-2, 100)
    sns.scatterplot(
        plot_temp,
        x="mean difference",
        y="-log10(q)",
        ax=ax,
        hue="hue",
        alpha=0.5,
        linewidth=0.05,
        palette={1: "#d62828", 0: "#9b9b98", -1: "#003049"},
    )

    ax.axhline(-np.log10(0.00001), zorder=0, c="#9b9b98", lw=1.2, ls="--", alpha=0.5)
    ax.axvline(0.1, zorder=0, c="#d62828", lw=1.2, ls="--", alpha=0.5)
    ax.axvline(-0.1, zorder=0, c="#003049", lw=1.2, ls="--", alpha=0.5)

    # ax.set(box_aspect=)
    ax.legend_ = None
    ax.set(title="%s - %s" % (group_A, group_B), ylabel="-log10(adjusted p)")
    # for i, point in C9_plot_temp.head(5).iterrows():
    #    ax.text(point['mean difference']-.1, point['-log10(q)']+2, str(point['annot.symbol']),size=10)

    texts = []
    for i, point in (
        plot_temp[
            (plot_temp["q_new"] < 1e-8)
            & (~plot_temp["annot.symbol"].str.startswith(("MIR", "LOC", "LINC")))
        ]
        .sort_values(by="mean difference")[
            ["mean difference", "-log10(q)", "annot.symbol"]
        ]
        .drop_duplicates()
        .head(5)
        .iterrows()
    ):
        texts.append(
            ax.text(
                point["mean difference"],
                point["-log10(q)"],
                str(point["annot.symbol"]),
                size=11,
            )
        )

    for i, point in (
        plot_temp[
            (plot_temp["q_new"] < 1e-8)
            & (~plot_temp["annot.symbol"].str.startswith(("MIR", "LOC", "LINC")))
        ]
        .sort_values(by="mean difference")[
            ["mean difference", "-log10(q)", "annot.symbol"]
        ]
        .drop_duplicates()
        .tail(5)
        .iterrows()
    ):
        texts.append(
            ax.text(
                point["mean difference"],
                point["-log10(q)"],
                str(point["annot.symbol"]),
                size=11,
            )
        )

    adjust_text(
        texts,
        arrowprops=dict(arrowstyle="-", color="grey", lw=0.7),
        expand_align=(0.5, 4),
        expand_text=(1, 4),
        force_text=(1, 3),
        precision=0.0001,
    )

    # f.savefig(
    #    './202401002_%s_%s_recalculate_MWUp_volvano.tiff'
    #    % (group_A, group_B),
    #    bbox_inches='tight', dpi=600)
```
